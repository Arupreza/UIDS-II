```python
def plot_comparison(df1, df2, label1, label2, column='Intra_ID_Time_Gap', max_rows=2000, title='Intra-ID Time Gap Comparison'):
    """
    Plots a specified column from two DataFrames on the same axes for comparison.

    Args:
        df1 (pd.DataFrame): The first DataFrame.
        df2 (pd.DataFrame): The second DataFrame.
        label1 (str): The label for the plot from the first DataFrame.
        label2 (str): The label for the plot from the second DataFrame.
        column (str): The name of the column to plot from each DataFrame.
        max_rows (int): The number of rows to include in the plot.
        title (str): The title for the chart.
    """
    # Set up the plot aesthetics
    plt.style.use('seaborn-v0_8-whitegrid')
    plt.figure(figsize=(14, 6))

    # Plot the specified column and slice from each DataFrame
    plt.plot(df1[column][:max_rows], label=label1, color='royalblue', linewidth=1.5)
    plt.plot(df2[column][:max_rows], label=label2, color='orangered', linestyle='--', linewidth=1.5)

    # Add informative labels and title
    plt.title(title, fontsize=16, weight='bold')
    plt.xlabel('Message Index', fontsize=12)
    plt.ylabel(column.replace('_', ' '), fontsize=12)
    plt.legend(fontsize=11)
    plt.tight_layout()
    
    # Display the plot
    plt.show()

#plot_comparison(Gen_AF, Gen_Rep_L, 'Kia_AF', 'Kia_DoS_H', column='Intra_ID_Time_Gap', max_rows=2000, title='Kia_AF vs Kia_DoS_H Time Delta Normalized Comparison')
```

```python
def SegmentFromFile(filename, time_gap):
    """
    Loads data from a file and segments it into time-based chunks.
    """
    # --- THIS IS THE FIX ---
    # 1. Define the base directory where all your CSV files are stored.
    DATA_DIRECTORY = "/home/lisa/Arupreza/UIDS-II/Input_data/"
    
    # 2. Construct the full, absolute path to the file.
    full_path = DATA_DIRECTORY + filename 
    # --- END OF FIX ---

    # --- Step 1: Load Data ---
    # The rest of your function can now use 'full_path'
    # and your LoadPreprocessData function will find the file.
    
    if not full_path.lower().endswith('.csv'):
        full_path_with_ext = f"{full_path}.csv"
    else:
        full_path_with_ext = full_path

    df = LoadPreprocessData(full_path_with_ext)

    if df is None:
        return [], []

    # --- Step 2: Segment Data ---
    # (The rest of the function remains exactly the same)
    df = df.sort_values('Time_Offset').reset_index(drop=True)
    df_subset = df[['Time_Offset', 'CAN_ID_Norm', 'Time_Delta_Norm', 'Intra_ID_Time_Gap_Norm', 'Label']].copy()

    chunks = []
    labels = []

    min_time = df_subset['Time_Offset'].min()
    max_time = df_subset['Time_Offset'].max()
    
    window_start = min_time

    while window_start <= max_time:
        window_end = window_start + time_gap
        segment_df = df_subset[(df_subset['Time_Offset'] >= window_start) & (df_subset['Time_Offset'] < window_end)]

        if not segment_df.empty:
            features = segment_df[['Time_Delta_Norm', 'Intra_ID_Time_Gap_Norm']].values
            chunks.append(features)
            is_attack = any(segment_df['Label'] != 'Normal')
            labels.append(1 if is_attack else 0)

        window_start += time_gap

    return chunks, labels
```

```python
def generate_attack_summary(attack_name, car_models, time_gap):
    """
    Generates a specific summary report for an attack type.
    
    The output format is fixed to match the user's request, including both
    the normal (AF) data and the L/M/H variants of the specified attack.
    
    Args:
        attack_name (str): The name of the attack (e.g., 'DoS').
        car_models (list): The list of car models (e.g., ['Kia', 'Tesla']).
        time_gap (int): The time window for segmentation.
    """
    # 1. Process and print the Ambient/Normal (AF) data first
    for car in car_models:
        filename = f"{car}_AF"
        # The comma in print() adds a space, so we use an f-string for exact formatting.
        segmented_data = SegmentFromFile(filename, time_gap)
        if segmented_data and segmented_data[0]:
            print(f'{car}: {len(segmented_data[0][0])}')
        else:
            print(f'{car}: 0')

    # 2. Process and print the different levels of the specified attack
    attack_levels = {'L': 'Low', 'M': 'Medium', 'H': 'High'}
    for level_code, level_name in attack_levels.items():
        # Print separators and the header for each attack level
        print('################################')
        print(f'{attack_name} {level_name}')
        print('################################')
        
        for car in car_models:
            filename = f"{car}_{attack_name}_{level_code}"
            segmented_data = SegmentFromFile(filename, time_gap)
            if segmented_data and segmented_data[0]:
                print(f'{car}: {len(segmented_data[0][0])}')
            else:
                print(f'{car}: 0')
```
