```python
def data_to_gasf(data, sample_range=(-1, 1)):
   """
   Convert multi-column data to GASF images.
   
   Parameters:
   - data: numpy array of shape (n_samples, n_features)
   - sample_range: tuple, rescaling range (default (-1,1) for GASF)
   
   Returns:
   - List of GASF matrices, one for each column
   """
   
   def rescale(x, sample_range):
       """Rescale data to sample_range"""
       x_min, x_max = np.min(x), np.max(x)
       if x_max == x_min:
           return np.zeros_like(x)
       return ((x - x_min) / (x_max - x_min)) * (sample_range[1] - sample_range[0]) + sample_range[0]
   
   def gasf(x):
       """Gramian Angular Summation Field"""
       # Rescale to [-1, 1]
       x_scaled = rescale(x, sample_range)
       # Clamp to avoid numerical issues with arccos
       x_scaled = np.clip(x_scaled, -1, 1)
       # Convert to angular representation
       phi = np.arccos(x_scaled)
       
       # Create GASF matrix
       n = len(phi)
       gasf_matrix = np.zeros((n, n))
       for i in range(n):
           for j in range(n):
               gasf_matrix[i, j] = np.cos(phi[i] + phi[j])
       
       return gasf_matrix
   
   # Process each column
   gasf_images = []
   for col_idx in range(data.shape[1]):
       column_data = data[:, col_idx]
       gasf_matrix = gasf(column_data)
       gasf_images.append(gasf_matrix)
   
   return gasf_images

def plot_gasf_images(gasf_images, column_names=None, figsize=(15, 5)):
   """Plot GASF images for visualization"""
   n_cols = len(gasf_images)
   
   if column_names is None:
       column_names = [f'Column_{i}' for i in range(n_cols)]
   
   fig, axes = plt.subplots(1, n_cols, figsize=figsize)
   if n_cols == 1:
       axes = [axes]
   
   for i, (gasf_img, name) in enumerate(zip(gasf_images, column_names)):
       im = axes[i].imshow(gasf_img, cmap='coolwarm', origin='upper')
       axes[i].set_title(f'GASF - {name}')
       axes[i].axis('off')
       plt.colorbar(im, ax=axes[i])
   
   plt.tight_layout()
   plt.show()

# Usage with your data
def process_can_data(data_array):
   """Process your CAN data to GASF images"""
   gasf_images = data_to_gasf(data_array)
   column_names = ['Cat_CAN_ID', 'Intra_ID_Time_Gap', 'Time_Gap']
   plot_gasf_images(gasf_images, column_names)
   return gasf_images
```
