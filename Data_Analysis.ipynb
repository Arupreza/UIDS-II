```python
Kia = ['0356', '0366', '0368', '0260', '0329', '0367'] #22813
Sil = ['018E', '00C9', '00D3', '00AA', '00BE', '01ED'] #18670
Tesla = ['0187', '0186', '02D5', '018A', '0256', '0142'] #19163
Gen = ['00B5', '0100', '007A', '00A5', '006A', '0040'] #23803

def load_data(df_name, IDs):

    from sklearn.preprocessing import MinMaxScaler
    scaler = MinMaxScaler()

    path = "/home/lisa/Arupreza/Cognitive-Belief-Driven-Q-Learning-for-Vehicle-Model-Agnostic-Intrusion-Detection-in-V-Net/Input_data/"
    df = pd.read_csv(path + df_name)
    df['Intra_ID_Time_Gap'] = (df.groupby('CAN_ID')['Time_Offset'].diff().fillna(-1))
    df = df[df["CAN_ID"].isin(IDs)]
    df = df[df['Intra_ID_Time_Gap'] != -1.0].reset_index(drop=True)
    df = df[['Intra_ID_Time_Gap']]
    df[['Intra_ID_Time_Gap']] = scaler.fit_transform(df[['Intra_ID_Time_Gap']])
    
    return df
```

```python
def segment_df(df, segment_size=30, drop_incomplete=True):
    arr = df[['Intra_ID_Time_Gap']].to_numpy()
    segments = []
    for start in range(0, arr.shape[0], segment_size):
        seg = arr[start : start + segment_size]
        if drop_incomplete and seg.shape[0] < segment_size:
            continue        # skip the short tail
        segments.append(seg)
    return segments
```

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_gadf(data_array):
    """
    Generate Gramian Angular Difference Field (GADF) plot.
    
    Parameters:
    data_array: numpy array of shape (n_samples, 1) or (n_samples,)
    
    Returns:
    gramian_field: numpy array, the Gramian Angular Difference Field
    """
    
    # Flatten if needed
    if data_array.ndim > 1:
        data_1d = data_array.flatten()
    else:
        data_1d = data_array
    
    # Normalize to [-1, 1] range
    data_min = np.min(data_1d)
    data_max = np.max(data_1d)
    data_scaled = 2 * (data_1d - data_min) / (data_max - data_min) - 1
    
    # Convert to polar coordinates (angular representation)
    # phi = arccos(x) where x is the scaled value
    phi = np.arccos(data_scaled)
    
    # Create time indices normalized to [0, 1]
    n = len(data_scaled)
    time_indices = np.linspace(0, 1, n)
    
    # Create Gramian Angular Difference Field
    # GADF: sin(phi_i + phi_j) 
    gramian_field = np.sin(phi[:, None] + phi[None, :])
    
    # Plot GADF only
    plt.figure(figsize=(8, 6))
    im = plt.imshow(gramian_field, cmap='RdBu_r', origin='lower', 
                    extent=[0, n-1, 0, n-1])
    plt.title('GADF Plot')
    plt.xlabel('Time Index')
    plt.ylabel('Time Index')
    plt.colorbar(im)
    plt.show()
    
    print(f"Data shape: {data_1d.shape}")
    print(f"Gramian field shape: {gramian_field.shape}")
    print(f"Value range: [{np.min(data_1d):.4f}, {np.max(data_1d):.4f}]")
    print(f"Scaled range: [{np.min(data_scaled):.4f}, {np.max(data_scaled):.4f}]")
    
    return gramian_field

# Example usage:
# data = np.array([[0.54545455], [0.4969697 ], [0.4969697 ], ...])
# gadf_field = generate_gadf(data)
```
