```python
def load_and_process(path, filename):
    file_path = os.path.join(path, filename)
    df = pd.read_csv(file_path, sep="\t", engine="python")
    df = df[2:][cols].reset_index(drop=True)
    df["Time_Gap"] = df["Time_Offset"].diff(1)
    return df
```
